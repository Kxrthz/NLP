### Aim:
To implement state-of-the-art token classification and Named Entity Recognition (NER) utilizing the pretrained transformer architecture `dslim/bert-base-NER` via Hugging Face `transformers` pipeline.

```python
from transformers import pipeline

print("Libraries imported successfully!")

sentences = [
    "Jensen Huang announced next generation GPUs at NVIDIA headquarters in Santa Clara.",
    "Demis Hassabis leads Google DeepMind artificial intelligence research in London.",
    "OpenAI was co-founded by Sam Altman in San Francisco.",
    "The international AI conference will take place in Tokyo on 15 September 2026."
]

print("\nInput Sentences:")
for i, sentence in enumerate(sentences, 1):
    print(f"S{i}: {sentence}")
```

```python
# Load pretrained BERT NER model
ner_model = pipeline(
    "ner",
    model="dslim/bert-base-NER",
    aggregation_strategy="first"
)

print("Pretrained BERT NER model loaded successfully!")
```

```python
print("Named Entity Recognition Results")
print("=" * 60)

for i, sentence in enumerate(sentences, 1):
    entities = ner_model(sentence)
    print(f"\nS{i}: {sentence}")

    for entity in entities:
        print(
            f"Entity: {entity['word']:<22} "
            f"Type: {entity['entity_group']:<10} "
            f"Score: {entity['score']:.4f}"
        )
```

```python
print("\n" + "=" * 60)
print("             BERT NAMED ENTITY RECOGNITION")
print("=" * 60)

for i, sentence in enumerate(sentences, 1):
    entities = ner_model(sentence)
    print(f"\nS{i}: {sentence}")
    print("-" * 60)

    for entity in entities:
        print(f"  [{entity['entity_group']}]  {entity['word']:<20} (Confidence: {entity['score']:.4f})")
print("=" * 60)
```

```python
from transformers import pipeline

print("Libraries imported successfully!")

# Post-Lab sentences
sentences = [
    "Tim Cook inaugurated the new Apple retail store in Mumbai.",
    "Amazon Web Services established a massive cloud datacenter in Frankfurt.",
    "ISRO successfully launched the lunar exploration probe from Sriharikota."
]

print("\nPost-Lab Input Sentences:")
for i, sentence in enumerate(sentences, 1):
    print(f"S{i}: {sentence}")
```

```python
# Load pretrained BERT NER model
ner_model = pipeline(
    "ner",
    model="dslim/bert-base-NER",
    aggregation_strategy="first"
)

print("Pretrained BERT NER model loaded successfully!")
```

```python
print("Named Entity Recognition Results")
print("=" * 60)

for i, sentence in enumerate(sentences, 1):
    entities = ner_model(sentence)
    print(f"\nS{i}: {sentence}")

    for entity in entities:
        print(
            f"Entity: {entity['word']:<24} "
            f"Type: {entity['entity_group']:<10} "
            f"Score: {entity['score']:.4f}"
        )
```

### Model Summary & Observations:
1. **Transformer Contextual Power**: Pretrained bidirectional transformer representations (`bert-base-NER`) leverage deep self-attention to accurately classify named entities across diverse domains without manual feature engineering.
2. **Entity Types Identified**:
   - **PER (Person)**: `Jensen Huang`, `Demis Hassabis`, `Sam Altman`, `Tim Cook`
   - **ORG (Organization)**: `NVIDIA`, `Google DeepMind`, `OpenAI`, `Apple`, `Amazon Web Services`, `ISRO`
   - **LOC (Location)**: `Santa Clara`, `London`, `San Francisco`, `Tokyo`, `Mumbai`, `Frankfurt`, `Sriharikota`
3. **Subword Aggregation**: Setting `aggregation_strategy="first"` reliably merges WordPiece subwords (e.g., multi-token entity strings) into unified human-readable entity spans with >99% confidence scores.
