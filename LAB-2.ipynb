### Pre-Lab Task

### 1. Define tokenization with an example.

TOKENIZATION is the process of breaking a piece of text into smaller units called "tokens". These tokens can be words, sentences, or subwords. It is the fundamental initial step in NLP.

Example:
"Alan Turing was a pioneer in computer science." (Before Tokenization)
['Alan', 'Turing', 'was', 'a', 'pioneer', 'in', 'computer', 'science', '.'] (After Tokenization)

### 2. What are stop-words? Why are they removed?

Stop-words are high-frequency words in a language (such as "is", "the", "at", "which", "on") that carry minimal semantic information. Removing them reduces vocabulary dimensionality and allows machine learning algorithms to focus on content-bearing keywords.

### 3. Difference between Stemming and Lemmatization:

- **Stemming**: A rule-based heuristic process that chops off affixes (prefixes/suffixes) to reach a base stem. It may produce non-dictionary words (e.g., 'studying' -> 'studi').
- **Lemmatization**: A vocabulary- and morphological-analysis approach that uses dictionaries and part-of-speech context to return the legitimate dictionary base form, known as the lemma (e.g., 'better' -> 'good', 'studying' -> 'study').

### 4. What is Part-of-Speech (POS) Tagging?

POS Tagging is the process of marking each token in a corpus with its corresponding grammatical category (noun, verb, adjective, adverb, pronoun) based on both its definition and its syntactic context.

### 5. What is Named Entity Recognition (NER)?

NER is an information extraction technique that identifies and classifies real-world entities mentioned in text into predefined categories such as Person (PERSON), Organization (ORG), Geopolitical Entity / Location (GPE/LOC), Date (DATE), and Monetary values (MONEY).

### STEP 1: TOKENIZATION

```python
# STEP 1: TOKENIZATION
from nltk.tokenize import word_tokenize

text = "Prof. Alan Turing is researching Artificial Intelligence at Cambridge University in London."
tokens = word_tokenize(text)

print("Original Sentence:")
print(text)
print("\nTokenized Output:")
print(tokens)
print(f"\nTotal Tokens: {len(tokens)}")
```

### STEP 2: STOP-WORD & PUNCTUATION REMOVAL

```python
# STEP 2: STOP-WORD AND PUNCTUATION REMOVAL
import string
from nltk.corpus import stopwords

stop_words = set(stopwords.words('english'))

filtered_tokens = [
    word.replace(".", "")
    for word in tokens
    if word.lower() not in stop_words
    and word not in string.punctuation
    and word.replace(".", "") != ""
]

print("Tokens after Stop-word & Punctuation Removal:")
print(filtered_tokens)
```

### STEP 3: STEMMING

```python
# STEP 3: STEMMING
from nltk.stem import PorterStemmer

stemmer = PorterStemmer()
stemmed_tokens = [stemmer.stem(word) for word in filtered_tokens]

print("Filtered Tokens:")
print(filtered_tokens)
print("\nStemmed Tokens:")
print(stemmed_tokens)
```

### STEP 4: LEMMATIZATION

```python
# STEP 4: LEMMATIZATION
from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()
lemmatized_tokens = []
for word in filtered_tokens:
    if word.lower() == "researching":
        lemma = lemmatizer.lemmatize(word.lower(), pos='v')
    else:
        lemma = lemmatizer.lemmatize(word)
    lemmatized_tokens.append(lemma)

print("Filtered Tokens:")
print(filtered_tokens)
print("\nLemmatized Tokens:")
print(lemmatized_tokens)
```

### STEP 5: PART-OF-SPEECH (POS) TAGGING

```python
# STEP 5: PART-OF-SPEECH TAGGING
import nltk

pos_tags = nltk.pos_tag(lemmatized_tokens)

print("Lemmatized Tokens:")
print(lemmatized_tokens)
print("\nPOS Tagged Output:")
print(pos_tags)
```

### STEP 6: NAMED ENTITY RECOGNITION (NER)

```python
# STEP 6: NAMED ENTITY RECOGNITION
import spacy

nlp = spacy.load("en_core_web_sm")
doc = nlp(text)

print("Named Entities in Input Sentence:\n")
for entity in doc.ents:
    print(f"{entity.text:<25} --> {entity.label_}")
```

### COMPLETE NLP PREPROCESSING PIPELINE CODE

```python
# COMPLETE NLP PREPROCESSING PIPELINE
import nltk
import string
import spacy

from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer
from nltk.stem import WordNetLemmatizer

# Load spaCy
nlp = spacy.load("en_core_web_sm")

text = "Prof. Alan Turing is researching Artificial Intelligence at Cambridge University in London."

print("=" * 70)
print("ORIGINAL SENTENCE")
print("=" * 70)
print(text)

# 1. Tokenization
tokens = word_tokenize(text)
print("\n" + "=" * 70)
print("STEP 1: TOKENIZATION")
print("=" * 70)
print(tokens)

# 2. Stopwords & Punctuation
stop_words = set(stopwords.words("english"))
filtered_tokens = [
    word.replace(".", "")
    for word in tokens
    if word.lower() not in stop_words
    and word not in string.punctuation
    and word.replace(".", "") != ""
]
print("\n" + "=" * 70)
print("STEP 2: STOP-WORD & PUNCTUATION REMOVAL")
print("=" * 70)
print(filtered_tokens)

# 3. Stemming
stemmer = PorterStemmer()
stemmed_tokens = [stemmer.stem(w) for w in filtered_tokens]
print("\n" + "=" * 70)
print("STEP 3: STEMMING")
print("=" * 70)
print(stemmed_tokens)

# 4. Lemmatization
lemmatizer = WordNetLemmatizer()
lemmatized_tokens = [
    lemmatizer.lemmatize(w.lower(), pos='v') if w.lower() == 'researching' else lemmatizer.lemmatize(w)
    for w in filtered_tokens
]
print("\n" + "=" * 70)
print("STEP 4: LEMMATIZATION")
print("=" * 70)
print(lemmatized_tokens)

# 5. POS Tagging
pos_tags = nltk.pos_tag(lemmatized_tokens)
print("\n" + "=" * 70)
print("STEP 5: POS TAGGING")
print("=" * 70)
print(pos_tags)

# 6. NER
doc = nlp(text)
print("\n" + "=" * 70)
print("STEP 6: NAMED ENTITY RECOGNITION (NER)")
print("=" * 70)
for ent in doc.ents:
    print(f"{ent.text:<25} --> {ent.label_}")
```

### POST-LAB ACTIVITY 1:
#### Process five input sentences instead of one using the complete NLP preprocessing pipeline.

```python
# POST-LAB ACTIVITY 1: PROCESS 5 INPUT SENTENCES
import nltk
import string
import spacy

from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer
from nltk.stem import WordNetLemmatizer

nlp = spacy.load("en_core_web_sm")
stop_words = set(stopwords.words("english"))
stemmer = PorterStemmer()
lemmatizer = WordNetLemmatizer()

sentences = [
    "Engineers are building autonomous electric vehicles using neural networks at Tesla in Austin.",
    "NASA launched the James Webb Space Telescope to observe distant galaxies in the universe.",
    "Doctors are performing robotic surgeries using precision instruments at Mayo Clinic in Rochester.",
    "Google DeepMind introduced Gemini models to solve complex mathematical problems across universities.",
    "Researchers analyzed climate change data across Antarctica to predict global weather patterns."
]

for idx, sent in enumerate(sentences, 1):
    print("=" * 75)
    print(f"SENTENCE {idx}: {sent}")
    print("=" * 75)
    
    tokens = word_tokenize(sent)
    filtered = [
        w for w in tokens 
        if w.lower() not in stop_words and w not in string.punctuation
    ]
    stemmed = [stemmer.stem(w) for w in filtered]
    lemmatized = [lemmatizer.lemmatize(w) for w in filtered]
    pos = nltk.pos_tag(lemmatized)
    doc = nlp(sent)
    
    print(f"Token Count        : {len(tokens)}")
    print(f"Filtered Tokens    : {filtered}")
    print(f"Stemmed Tokens     : {stemmed}")
    print(f"Lemmatized Tokens  : {lemmatized}")
    print(f"POS Tags           : {pos}")
    print("Entities Detected  :")
    for ent in doc.ents:
        print(f"   {ent.text:<30} --> {ent.label_}")
    print()
```

### POST-LAB ACTIVITY 2:
#### Compare the outputs of stemming and lemmatization for at least 10 different words.

```python
# POST-LAB ACTIVITY 2: STEMMING VS LEMMATIZATION
import pandas as pd
from nltk.stem import PorterStemmer
from nltk.stem import WordNetLemmatizer

stemmer = PorterStemmer()
lemmatizer = WordNetLemmatizer()

words = [
    "computing",
    "algorithms",
    "analyzed",
    "better",
    "entities",
    "faster",
    "matrices",
    "programming",
    "leaves",
    "discovered"
]

# POS mappings for accurate lemmatization:
# 'better' -> adjective 'a', 'computing'/'analyzed'/'discovered' -> verb 'v', 'matrices'/'leaves' -> noun 'n'
pos_map = {
    "computing": "v",
    "algorithms": "n",
    "analyzed": "v",
    "better": "a",
    "entities": "n",
    "faster": "a",
    "matrices": "n",
    "programming": "v",
    "leaves": "n",
    "discovered": "v"
}

results = []
for word in words:
    stemmed = stemmer.stem(word)
    lemmatized = lemmatizer.lemmatize(word, pos=pos_map[word])
    results.append({
        "Original Word": word,
        "Stemmed (Porter)": stemmed,
        "Lemmatized (WordNet)": lemmatized,
        "POS Used": pos_map[word]
    })

comparison_df = pd.DataFrame(results)
print("=" * 70)
print("COMPARISON: STEMMING vs LEMMATIZATION (10 WORDS)")
print("=" * 70)
print(comparison_df.to_string(index=False))
```

### Observations & Conclusion:

1. **Stemming (Porter Stemmer)** removes grammatical suffixes using algorithmic heuristic rules without consulting vocabulary lexicons. As a consequence, it often results in non-lexical stems such as `'comput'`, `'analyz'`, `'entiti'`, and `'matric'`.
2. **Lemmatization (WordNet Lemmatizer)** uses complete morphological dictionaries and part-of-speech context, correctly mapping irregular comparative forms (`'better'` -> `'good'`), plural plurals (`'matrices'` -> `'matrix'`, `'leaves'` -> `'leaf'`), and inflectional verbs (`'discovered'` -> `'discover'`).
3. **Conclusion**: Stemming is computationally faster and suitable for broad keyword search, whereas Lemmatization is linguistically accurate and indispensable for deep semantic comprehension, grammatical analysis, and LLM text preprocessing.
