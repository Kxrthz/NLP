```python
import tensorflow as tf

print("TensorFlow version:", tf.__version__)
```

```python
import numpy as np
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense

print("Libraries imported successfully!")

# Hospitality & Hotel guest reviews dataset
reviews = [
    "The hotel suite was magnificent and breakfast was delicious",
    "Staff members were extraordinarily courteous and helpful",
    "Luxurious swimming pool with stunning ocean views",
    "Delightful stay with prompt room service and spotless amenities",
    "The hotel room was dirty and air conditioner was noisy",
    "Terrible front desk experience and extremely rude management",
    "Unpleasant smell in bathroom and poor housekeeping quality",
    "Very disappointing vacation with delayed check in and broken shower"
]

# Sentiment labels: 1 = Positive, 0 = Negative
labels = np.array([1, 1, 1, 1, 0, 0, 0, 0])

print("\nGuest Reviews:")
for i, review in enumerate(reviews, 1):
    sentiment = "Positive" if labels[i-1] == 1 else "Negative"
    print(f"{i}. {review} -> {sentiment}")
```

```python
# Tokenization
tokenizer = Tokenizer()
tokenizer.fit_on_texts(reviews)

sequences = tokenizer.texts_to_sequences(reviews)

print("Word Index:")
print(tokenizer.word_index)

print("\nInteger Encoded Sentences:")
for i, seq in enumerate(sequences, 1):
    print(f"{i}: {seq}")
```

```python
# Padding
max_length = max(len(seq) for seq in sequences)

X = pad_sequences(
    sequences,
    maxlen=max_length,
    padding="post"
)

print("Padded Sequences:")
print(X)

print("\nSequence Shape:", X.shape)
print("Maximum Sequence Length:", max_length)
```

```python
# Build LSTM Model Architecture
vocab_size = len(tokenizer.word_index) + 1
embedding_dim = 16

model = Sequential([
    Embedding(vocab_size, embedding_dim),
    LSTM(32),
    Dense(1, activation="sigmoid")
])

model.compile(
    loss="binary_crossentropy",
    optimizer="adam",
    metrics=["accuracy"]
)

model.summary()
```

```python
# Train the LSTM Model
history = model.fit(
    X,
    labels,
    epochs=100,
    verbose=0
)

print("LSTM model trained successfully for 100 epochs!")
```

```python
# Predict Sentiment on Reviews
predictions = model.predict(X, verbose=0)

print("Sentiment Predictions:")
print("=" * 60)

for i, probability in enumerate(predictions, 1):
    sentiment = "Positive" if probability[0] >= 0.5 else "Negative"
    actual = "Positive" if labels[i-1] == 1 else "Negative"
    print(f"Review {i}: {sentiment:<10} (Prob: {probability[0]:.4f}) | Actual: {actual}")
```

```python
# Test Inference Function on New Review
def predict_sentiment(review_text):
    seq = tokenizer.texts_to_sequences([review_text])
    padded = pad_sequences(seq, maxlen=max_length, padding="post")
    prob = model.predict(padded, verbose=0)[0][0]
    sentiment = "Positive" if prob >= 0.5 else "Negative"
    return sentiment, prob

test_reviews = [
    "The restaurant dining was exquisite and staff was charming",
    "Horrible dirty room and terrible air conditioning"
]

print("Inference on Unseen Test Reviews:")
print("=" * 60)
for rev in test_reviews:
    pred, p = predict_sentiment(rev)
    print(f"Review : {rev}")
    print(f"Result : {pred} (Confidence: {p:.4f})")
    print("-" * 60)
```

### Architecture Insights & Conclusion:
1. **Embedding Layer**: Transforms discrete word integers into continuous 16-dimensional dense embedding vectors.
2. **LSTM Recurrent Cell**: Captures sequential order and temporal dependencies across token sequences using input, forget, and output gates, avoiding the vanishing gradient problem.
3. **Dense Output Unit**: Employs a sigmoid activation function $\sigma(z) = rac{1}{1 + e^{-z}}$ to yield calibrated posterior class probability $P(y=1|X) \in [0, 1]$.
4. The model attains high training discrimination across positive and negative sentiment sequences.
