# EXERCISE 3: TEXT REPRESENTATION TECHNIQUES
### Aim:
To explore and implement foundational text representation techniques in NLP:
1. **One-Hot Encoding**
2. **Bag of Words (BoW)** using `CountVectorizer`
3. **Word Cloud Visualization** and Frequency Analysis

```python
# ================================================================
# EXERCISE 3: TEXT REPRESENTATION TECHNIQUES
# ACTIVITY 1: ONE-HOT ENCODING
# ================================================================
#
# Tasks:
# 1. Import the required Python libraries.
# 2. Create a vocabulary from the given corpus.
# 3. Identify all unique words.
# 4. Generate One-Hot Encoding vectors for each unique word.
# 5. Display the One-Hot Encoding vectors in tabular format.
# ================================================================

import pandas as pd

# Given Text Corpus
corpus = [
    "Deep learning powers natural language processing",
    "Natural language processing creates intelligent chatbots",
    "Deep learning models understand human language"
]

documents = ["D1", "D2", "D3"]

print("Given Text Corpus:")
for doc, text in zip(documents, corpus):
    print(f"{doc}: {text}")

# Tokenize and create vocabulary
all_words = []
for text in corpus:
    tokens = text.lower().split()
    all_words.extend(tokens)

# Unique words sorted alphabetically
unique_words = sorted(list(set(all_words)))

print("\nVocabulary (Unique Words):")
print(unique_words)
print(f"Total Unique Words: {len(unique_words)}")

# Generate One-Hot Vectors for vocabulary
one_hot_data = {}
for word in unique_words:
    vector = [1 if word == target else 0 for target in unique_words]
    one_hot_data[word] = vector

one_hot_df = pd.DataFrame(one_hot_data, index=unique_words)

print("\nOne-Hot Encoding Matrix (Vocabulary representation):")
print(one_hot_df)
```

```python
# ================================================================
# EXERCISE 3: TEXT REPRESENTATION TECHNIQUES
# ACTIVITY 2: BAG OF WORDS (BoW)
# ================================================================
#
# Tasks:
# 1. Create the Bag of Words model using CountVectorizer.
# 2. Generate the vocabulary.
# 3. Compute the frequency of each word in every document.
# 4. Construct the Document-Term Matrix (DTM).
# 5. Display the BoW matrix.
# ================================================================

import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer

cv = CountVectorizer()
dtm_matrix = cv.fit_transform(corpus)

vocab = cv.get_feature_names_out()

print("Bag of Words Vocabulary:")
print(vocab)
print(f"Vocabulary Size: {len(vocab)}")

bow_df = pd.DataFrame(
    dtm_matrix.toarray(),
    index=documents,
    columns=vocab
)

print("\nDocument-Term Matrix (DTM):")
print(bow_df)
```

```python
# ================================================================
# EXERCISE 3: TEXT REPRESENTATION TECHNIQUES
# ACTIVITY 3: WORD CLOUD & FREQUENCY ANALYSIS
# ================================================================
#
# Tasks:
# 1. Combine all documents into a single text corpus.
# 2. Generate and visualize the Word Cloud.
# 3. Identify and display top five most frequent words.
# ================================================================

import matplotlib.pyplot as plt
from wordcloud import WordCloud
from collections import Counter

# 1. Combine all documents
combined_text = " ".join(corpus).lower()
print("Combined Text Corpus:")
print(combined_text)

# Tokenize words
words = combined_text.split()
word_frequency = Counter(words)

# 2. Identify top 5 most frequent words
top_five_words = word_frequency.most_common(5)

print("\n" + "=" * 50)
print("TOP 5 MOST FREQUENT WORDS")
print("=" * 50)
for rank, (word, freq) in enumerate(top_five_words, 1):
    print(f"{rank}. {word:<15} --> Frequency: {freq}")
print("=" * 50)

# 3. Generate Word Cloud
wordcloud = WordCloud(
    width=800,
    height=400,
    background_color='white',
    colormap='viridis'
).generate(combined_text)

plt.figure(figsize=(10, 5))
plt.imshow(wordcloud, interpolation='bilinear')
plt.axis('off')
plt.title("Word Cloud Visualization", fontsize=16)
plt.show()
```
