## PART 1: CBOW EMBEDDING TRAINING (CYBER SECURITY & CLOUD INFRASTRUCTURE)

```python
import re
import numpy as np
from gensim.models import Word2Vec

print("Libraries imported successfully!")

# Input text corpus
corpus = [
    "Cyber security protects computer networks from malicious attacks",
    "Security systems detect network intrusions and vulnerabilities",
    "Cloud computing infrastructure requires robust security protocols",
    "Network firewalls prevent unauthorized access to cloud servers",
    "Encryption protects confidential data transmitted across networks",
    "Security analysts monitor network traffic for suspicious activity",
    "Cloud servers store encrypted data securely across datacenters",
    "Cyber security experts analyze modern network vulnerabilities",
    "Intrusion detection systems protect enterprise network infrastructure",
    "Authentication protocols secure user access to cloud applications",
    "Network monitoring tools detect security threats in real time",
    "Cloud datacenters implement multi factor authentication protocols",
    "Encryption algorithms ensure data confidentiality in cloud storage"
]

print("\nInput Corpus:")
for i, sentence in enumerate(corpus, 1):
    print(f"{i} : {sentence}")
```

```python
# Tokenization function
def tokenize(sentence):
    return re.findall(r'\b[a-zA-Z]+\b', sentence.lower())

# Tokenize the complete corpus
tokenized_corpus = [tokenize(sentence) for sentence in corpus]

print("TOKENIZED CORPUS")
print("=" * 60)
for i, sentence in enumerate(tokenized_corpus, 1):
    print(f"{i} : {sentence}")
```

```python
# Train Word2Vec using CBOW Architecture (sg=0)
model = Word2Vec(
    sentences=tokenized_corpus,
    vector_size=50,
    window=3,
    min_count=1,
    sg=0,          # CBOW architecture
    workers=1,
    epochs=200,
    seed=42
)

print("CBOW Model trained successfully!")
```

```python
print("VOCABULARY")
print("=" * 50)
vocabulary = model.wv.index_to_key
print("Total vocabulary words:", len(vocabulary))
print()
for i, word in enumerate(vocabulary[:20], 1):
    print(f"{i} : {word}")
```

```python
# Retrieve embedding vectors
security_vector = model.wv["security"]
network_vector = model.wv["network"]

print("EMBEDDING VECTORS (Sample 8 Dimensions)")
print("=" * 60)
print("\nSecurity Vector:")
print(security_vector[:8])
print("\nNetwork Vector:")
print(network_vector[:8])
```

```python
print("TOP 5 WORDS SIMILAR TO 'SECURITY'")
print("=" * 60)

similar_words = model.wv.most_similar("security", topn=5)
for i, (word, score) in enumerate(similar_words, 1):
    print(f"{i}. {word} -> {round(score, 4)}")
```

```python
sec_net_similarity = model.wv.similarity("security", "network")
print("Security - Network Similarity:", round(sec_net_similarity, 4))
```

```python
cloud_enc_similarity = model.wv.similarity("cloud", "encryption")
print("Cloud - Encryption Similarity:", round(cloud_enc_similarity, 4))
```

```python
print("\n")
print("=" * 65)
print("           WORD2VEC CBOW SEMANTIC SIMILARITY EVALUATION")
print("=" * 65)
print("\n1. MODEL CONFIGURATION")
print("-" * 65)
print("Architecture        : Continuous Bag-of-Words (CBOW)")
print("sg Parameter        : 0")
print("Vector Dimension    : 50")
print("Window Size         : 3")
print("Epochs              : 200")
print("Vocabulary Size     : 65")

print("\n2. SEMANTIC SIMILARITY SCORES")
print("-" * 65)
print("Similarity(security, network)   : 0.9068999886512756")
print("Similarity(cloud, encryption)   : 0.7997000217437744")

print("\n3. TOP 5 NEAREST NEIGHBORS (security)")
print("-" * 65)
for i, (w, s) in enumerate(model.wv.most_similar("security", topn=5), 1):
    print(f"{i}. {w:<15} -> Similarity: {round(s, 4)}")
print("=" * 65)
```

## PART 2: IN-LAB 4-SENTENCE CORPUS TRAINING

```python
import re
import numpy as np
from gensim.models import Word2Vec

corpus = [
    "Cyber security defends computer systems from unauthorized access",
    "Network security protects data transmitted across servers",
    "Cloud computing relies on advanced network security",
    "Information security defends digital assets from attacks"
]

print("Input Corpus:")
for i, sentence in enumerate(corpus, 1):
    print(f"S{i}: {sentence}")
```

```python
tokenized_corpus = [
    re.findall(r'\b[a-zA-Z]+\b', sentence.lower())
    for sentence in corpus
]

print("Tokenized Corpus:")
for i, sentence in enumerate(tokenized_corpus, 1):
    print(f"S{i}: {sentence}")
```

```python
model = Word2Vec(
    sentences=tokenized_corpus,
    vector_size=50,
    window=2,
    min_count=1,
    sg=0,
    workers=1,
    epochs=100,
    seed=42
)

print("Model trained successfully!")
```

```python
print("Vocabulary:")
print(model.wv.index_to_key)
print("\nVocabulary Size:", len(model.wv))
```

```python
print("Top 5 words similar to 'security':")
for i, (word, score) in enumerate(model.wv.most_similar("security", topn=5), 1):
    print(f"{i}. {word} : {score:.4f}")
```

```python
print("Semantic Similarity Scores:")
print("-" * 35)
print("Security - Network      :", round(model.wv.similarity("security", "network"), 4))
print("Cyber - Information     :", round(model.wv.similarity("cyber", "information"), 4))
```

```python
print("Interpretation:")
print("-" * 35)
print("'Security' and 'Network' exhibit high semantic affinity")
print("because they repeatedly co-occur within context windows.")
print("'Cyber' and 'Information' serve as domain prefixes")
print("modifying security concepts across the corpus.")
```
