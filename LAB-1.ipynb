### STEP-BY-STEP IMPLEMENTATION

```python
# CREATE A PYTHON LIST CONTAINING THE GIVEN TEXT CORPUS

corpus = [
    "<p>Artificial Intelligence & Deep Learning are REVOLUTIONARY!!! Visit https://deeplearning.ai</p>",
    "PyTorch 2.5 and Transformers 5.0 are released in 2026.",
    "Large Language Models transform modern NLP.",
    "Contact our research lab at research-support@aistudio.org"
]
```

```python
# IMPORT THE REQUIRED LIBRARIES (re and BeautifulSoup)
import re
from bs4 import BeautifulSoup
```

```python
# CONVERT ALL TEXT TO LOWER CASE

for text in corpus:
    text = text.lower()
    print(text)
```

```python
lowercase_corpus = [text.lower() for text in corpus]
print(lowercase_corpus)
```

```python
# REMOVE HTML TAGS

html_removed = [
    BeautifulSoup(text, "html.parser").get_text()
    for text in lowercase_corpus
]
print(html_removed)
```

```python
# REMOVE URL'S

url_removed = [
    re.sub(r'https?://\S+|www\.\S+', '', text)
    for text in html_removed
]
print(url_removed)
```

```python
# REMOVE EMAIL ADDRESSES

email_removed = [
    re.sub(r'[\w\.-]+@[\w\.-]+\.\w+', '', text)
    for text in url_removed
]
print(email_removed)
```

```python
# REMOVE DIGITS

digit_removed = [
    re.sub(r'\d+', '', text)
    for text in email_removed
]
print(digit_removed)
```

```python
# REMOVE PUNCTUATION AND SPECIAL CHARACTERS

punct_removed = [
    re.sub(r'[^\w\s]', '', text)
    for text in digit_removed
]
print(punct_removed)
```

```python
# REMOVE EXTRA WHITE SPACES

cleaned_corpus = [
    re.sub(r'\s+', ' ', text).strip()
    for text in punct_removed
]
print(cleaned_corpus)
```

```python
# STORE THE CLEANED TEXT IN A NEW LIST NAMED cleaned_corpus

print("The cleaned text has been stored in the list 'cleaned_corpus'.")
print(cleaned_corpus)
```

```python
# DISPLAY THE ORIGINAL CORPUS AND THE CLEANED CORPUS

print("Original Corpus:\n")
for sentence in corpus:
    print(sentence)

print("\nCleaned Corpus:\n")
for sentence in cleaned_corpus:
    print(sentence)
```

```python
# COMPARE THE TEXT BEFORE AND AFTER CLEANING

print("Comparison Before and After Cleaning:\n")

for original, cleaned in zip(corpus, cleaned_corpus):
    print("Original :", original)
    print("Cleaned  :", cleaned)
    print("-" * 60)
```

```python
# FULL CODE PIPELINE

corpus = [
    "<p>Artificial Intelligence & Deep Learning are REVOLUTIONARY!!! Visit https://deeplearning.ai</p>",
    "PyTorch 2.5 and Transformers 5.0 are released in 2026.",
    "Large Language Models transform modern NLP.",
    "Contact our research lab at research-support@aistudio.org"
]

import re
from bs4 import BeautifulSoup

lowercase_corpus = [text.lower() for text in corpus]

print("=" * 70)
print("Step 3: Lowercase Conversion")
print("=" * 70)
for text in lowercase_corpus:
    print(text)

html_removed = [
    BeautifulSoup(text, "html.parser").get_text()
    for text in lowercase_corpus
]

print("\n" + "=" * 70)
print("Step 4: HTML Tags Removed")
print("=" * 70)
for text in html_removed:
    print(text)

url_removed = [
    re.sub(r'https?://\S+|www\.\S+', '', text)
    for text in html_removed
]

print("\n" + "=" * 70)
print("Step 5: URLs Removed")
print("=" * 70)
for text in url_removed:
    print(text)

email_removed = [
    re.sub(r'[\w\.-]+@[\w\.-]+\.\w+', '', text)
    for text in url_removed
]

print("\n" + "=" * 70)
print("Step 6: Email Addresses Removed")
print("=" * 70)
for text in email_removed:
    print(text)

digit_removed = [
    re.sub(r'\d+', '', text)
    for text in email_removed
]

print("\n" + "=" * 70)
print("Step 7: Digits Removed")
print("=" * 70)
for text in digit_removed:
    print(text)

punct_removed = [
    re.sub(r'[^\w\s]', '', text)
    for text in digit_removed
]

print("\n" + "=" * 70)
print("Step 8: Punctuation and Special Characters Removed")
print("=" * 70)
for text in punct_removed:
    print(text)

cleaned_corpus = [
    re.sub(r'\s+', ' ', text).strip()
    for text in punct_removed
]

print("\n" + "=" * 70)
print("Step 9: Extra White Spaces Removed")
print("=" * 70)
for text in cleaned_corpus:
    print(text)
```

```python
# POST-LAB DATASET: CUSTOMER REVIEWS

reviews = [
    "<div>Unbelievable gaming headset!!! 🎧 Sound immersion is 10/10. Check https://hypergear.io</div>",
    "Worst customer service ever!! 😡 Waited 45 mins in 2026. Email escalation@supportcare.net",
    "Superfast cloud deployment 🚀 and ultra-low latency! Highly recommended for developers.",
    "<p>Battery life on this smartwatch is INCREDIBLE ⚡ 5 days on single charge! Visit www.techpulse.com/smartwatch</p>",
    "Average build quality 📦 Delivery delayed by 4 days. Contact tracking-team@fastcargo.com for refund."
]

print("Original Reviews:\n")
for review in reviews:
    print(review)
```

```python
# CLEAN THE DATA / REVIEWS
cleaned_reviews = []

for review in reviews:
    # Convert to lowercase
    review = review.lower()

    # Remove HTML Tags
    review = BeautifulSoup(review, "html.parser").get_text()

    # Remove URLs
    review = re.sub(r'https?://\S+|www\.\S+', ' ', review)

    # Remove Email Addresses
    review = re.sub(r'[\w\.-]+@[\w\.-]+\.\w+', ' ', review)

    # Remove Digits
    review = re.sub(r'\d+', ' ', review)

    # Remove Punctuation and Special Characters / Emojis
    review = re.sub(r'[^\w\s]', ' ', review)

    # Remove Extra Spaces
    review = re.sub(r'\s+', ' ', review).strip()

    cleaned_reviews.append(review)
```

```python
# COMPARE BEFORE AND AFTER CLEANING

print("Comparison of Original and Cleaned Reviews:\n")

for original, cleaned in zip(reviews, cleaned_reviews):
    print("Original Review:")
    print(original)
    print("\nCleaned Review:")
    print(cleaned)
    print("=" * 60)
```

## REPORT ON TEXT CLEANING

Raw text scraped from websites and social media platforms frequently contains extraneous elements such as HTML markup, hyperlinks, contact email addresses, numeric codes, special symbols, and emojis. In natural language processing, such noisy elements dilute feature representations and harm machine learning performance.

Through the text cleaning pipeline implemented in this experiment:
1. **Case Normalization** unified vocabulary representation, preventing identical words from being treated as distinct tokens.
2. **Tag & Link Stripping** eliminated syntactical artifacts that provide zero contextual sentiment.
3. **Regex Punctuation Removal** removed noise while retaining essential linguistic tokens.
4. **Whitespace Condensation** standardized spacing across documents.

The resulting normalized text forms an optimal, noise-free foundation for downstream classification, sentiment evaluation, and LLM fine-tuning tasks.
