# EXERCISE 6: WORD EMBEDDINGS USING WORD2VEC
### Aim:
To train a Word2Vec continuous vector space representation model using Gensim, extract dense word embeddings, compute average sentence embeddings, and evaluate semantic cosine similarity between sentences.

```python
import re
import numpy as np
from gensim.models import Word2Vec
from sklearn.metrics.pairwise import cosine_similarity

print("Libraries imported successfully!")
```

```python
# Input Text Corpus
corpus = [
    "Autonomous robots navigate planetary surfaces",
    "Intelligent rovers explore cosmic missions",
    "Dolphins swim deep oceans",
    "Whales swim deep oceans"
]

print("Input Corpus:")
for i, sentence in enumerate(corpus, 1):
    print(f"S{i}: {sentence}")
```

```python
# Tokenization
tokenized_corpus = [
    re.findall(r'\b[a-zA-Z]+\b', sentence.lower())
    for sentence in corpus
]

print("Tokenized Corpus:")
for i, sentence in enumerate(tokenized_corpus, 1):
    print(f"S{i}: {sentence}")
```

```python
# Train Word2Vec Model
model = Word2Vec(
    sentences=tokenized_corpus,
    vector_size=50,
    window=3,
    min_count=1,
    sg=0,
    workers=1,
    epochs=300,
    seed=42
)

print("Word2Vec model trained successfully!")
print("Vocabulary words:", list(model.wv.index_to_key))
```

```python
print("Word Vectors (Sample):")
for word in ["robots", "navigate", "rovers", "explore", "dolphins", "oceans"]:
    print(f"\n{word} vector (first 8 dimensions):")
    print(model.wv[word][:8])
```

```python
# Function to create sentence embedding by averaging word vectors
def sentence_embedding(sentence):
    words = re.findall(r'\b[a-zA-Z]+\b', sentence.lower())
    vectors = [model.wv[word] for word in words if word in model.wv]
    return np.mean(vectors, axis=0)

# Create sentence embeddings
sentence1 = sentence_embedding("Autonomous robots navigate planetary surfaces")
sentence2 = sentence_embedding("Intelligent rovers explore cosmic missions")
sentence3 = sentence_embedding("Dolphins swim deep oceans")

print("Sentence 1 Embedding (first 8 dims):", sentence1[:8])
print("Sentence 2 Embedding (first 8 dims):", sentence2[:8])
print("Sentence 3 Embedding (first 8 dims):", sentence3[:8])
print("\nEmbedding Dimension:", sentence1.shape)
```

```python
# Compute Cosine Similarity
sim_1_2 = cosine_similarity(sentence1.reshape(1, -1), sentence2.reshape(1, -1))[0][0]
sim_1_3 = cosine_similarity(sentence1.reshape(1, -1), sentence3.reshape(1, -1))[0][0]

print("Cosine Similarity between S1 and S2 (Space/Robotics):", round(sim_1_2, 4))
print("Cosine Similarity between S1 and S3 (Robotics vs Marine):", round(sim_1_3, 4))
```

```python
print("=" * 65)
print("             WORD2VEC SENTENCE SIMILARITY EVALUATION")
print("=" * 65)
print(f"Sentence 1 : {corpus[0]}")
print(f"Sentence 2 : {corpus[1]}")
print("-" * 65)
print(f"Vector Dimension : {sentence1.shape[0]}")
print(f"Cosine Similarity: {round(sim_1_2, 4)}")
print("=" * 65)
if sim_1_2 >= 0.70:
    print("Result : High Semantic Relatedness")
elif sim_1_2 >= 0.40:
    print("Result : Moderate Semantic Relatedness")
else:
    print("Result : Low Semantic Relatedness")
print("=" * 65)
```
